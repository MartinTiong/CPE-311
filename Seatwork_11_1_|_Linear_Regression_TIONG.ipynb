PART 1

Suppose we had a linear regression model with one feature that learned the following bias and weights:

b=1

W1 =3

x1=0

```python
import numpy as np
import matplotlib.pyplot as plt

b = 1
w1 = 3
x1 = 0

y = b + w1 * x1
print(y)

xvalue = np.linspace(-5, 5, 100)
yvalue = b + w1 * xvalue

plt.figure()
plt.plot(xvalue, yvalue, label="y = 1 + 3x")

plt.scatter(x1, y)
plt.text(x1, y, "(0,1)")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Part 1")
plt.legend()
plt.show()
```
