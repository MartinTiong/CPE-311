```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("bhavikjikadara/mental-health-dataset")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd

# Check files inside dataset
print(os.listdir(path))
```

```python
df = pd.read_csv('/content/Mental Health Dataset.csv')
df.head()
```

3. Prepare the data by applying appropriate data preprocessing techniques.

```python
df.isnull().sum()

df = df.drop_duplicates()

df['Timestamp'] = pd.to_datetime(df['Timestamp'], errors='coerce')
df.dtypes
```

4. Analyze the data using descriptive analysis.


It shows that the dataset contains about 290,000 responses with 17 variables related to mental health. Most respondents were male and not self employed. Many individuals reported experiencing growing stress, mood swings, and coping struggles.

5. Perform correlation analysis.

```python
df_encoded = df.copy()

for col in df_encoded.select_dtypes(include=['object']).columns:
    df_encoded[col] = df_encoded[col].astype('category').cat.codes

correlation_matrix = df_encoded.corr()

correlation_matrix
```

6. Interpret the results based on the descriptive and correlation analysis.


The correlation analysis showed that therea re several relationships between variables. Individuals with a family history of mental illness were more likely to seek treatment. Growing stress showed a positive relationship with mood swings, suggesting that higher stress levels affect emotional stability. Coping struggles were negatively related to work interest, indicating that mental health challenges may reduce productivity.
